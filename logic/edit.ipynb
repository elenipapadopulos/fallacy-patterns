```python
import json
```

```python
with open("processed_syntax_examples_roberta-large.json", "r") as f:
    data = json.load(f)
```

```python
new_dict = {}
for definition, examples in data.items():
    new_dict[definition] = []
    for example in examples:
        new_dict[definition].append({
            "text": example[0],
            "label": example[1],
        })
```

```python
with open("syntax_examples_roberta.json", "w") as f:
    json.dump(new_dict, f, indent=4)
```
